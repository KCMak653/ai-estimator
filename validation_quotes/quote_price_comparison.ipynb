# Quote price comparison

Compare quote unit prices to pricebook (extracted units from flattened quotes).

```python
import pandas as pd
from pathlib import Path

CSV_PATH = Path("extracted_units_2026.csv")
df_2026 = pd.read_csv(CSV_PATH)
df_2026
```

```python
df_2026 = df_2026[(df_2026["color_in"] == "WHT") & (df_2026["color_out"] == "WHT")]
df_2026
df_2026.groupby(["window_description", "sq_ft_bucket"]).agg(
    count=("quote_unit_price", "count"),
    mean_quote=("quote_unit_price", "mean")
)
    
```
